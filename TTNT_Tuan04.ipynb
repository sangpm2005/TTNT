```python
#Bài mẫu
# Ma trận kề của đồ thị 6 đỉnh
G = [
    [0, 1, 1, 0, 1, 0],   # hàng của đỉnh A
    [1, 0, 1, 1, 0, 1],   # hàng của đỉnh B
    [1, 1, 0, 1, 1, 0],   # hàng của đỉnh C
    [0, 1, 1, 0, 0, 1],   # hàng của đỉnh D
    [1, 0, 1, 0, 0, 1],   # hàng của đỉnh E
    [0, 1, 0, 1, 1, 0]    # hàng của đỉnh F
]

# Tên các đỉnh của đồ thị (mỗi ký tự tương ứng một đỉnh)
node = "ABCDEF"

# t_ là bảng ánh xạ từ tên đỉnh -> chỉ số (ví dụ t_['A'] = 0)
t_ = {}
for i in range(len(G)):
    t_[node[i]] = i   # lưu vị trí i cho đỉnh node[i]

# Bậc (degree) của các đỉnh: tổng các phần tử trong hàng tương ứng
degree = []
for i in range(len(G)):
    degree.append(sum(G[i]))   # sum(G[i]) là số cạnh kề với đỉnh i

# Màu có thể sử dụng để tô cho các đỉnh (mỗi đỉnh bắt đầu có cùng một danh sách màu)
colorDict = {}
for i in range(len(G)):
    # gán cho mỗi đỉnh một bản sao danh sách màu khả dụng
    colorDict[node[i]] = ["Blue", "Red", "Yellow", "Green"]

# Sắp xếp các đỉnh theo thứ tự bậc giảm dần (dùng selection-like selection)
sortedNode = []   # danh sách đỉnh đã sắp xếp theo bậc giảm dần
indeks = []       # danh sách chỉ số đã được chọn

for i in range(len(degree)):
    _max = -1     # khởi tạo _max nhỏ hơn mọi bậc có thể (để tìm max)
    idx = None    # idx sẽ chứa chỉ số có bậc lớn nhất (trong vòng chưa chọn)
    for j in range(len(degree)):
        # nếu j chưa được chọn và degree[j] lớn hơn _max thì cập nhật
        if j not in indeks:
            if degree[j] > _max:
                _max = degree[j]
                idx = j
    # thêm chỉ số tìm được vào danh sách đã chọn
    indeks.append(idx)
    # thêm tên đỉnh tương ứng vào sortedNode
    sortedNode.append(node[idx])

# Phần xử lý tô màu (thuật toán tham lam: với mỗi đỉnh theo sortedNode, gán màu đầu tiên khả dụng)
theSolution = {}   # lưu kết quả tô màu: {tên đỉnh: màu}

for n in sortedNode:
    # danh sách màu khả dụng hiện tại cho đỉnh n
    available_colors = colorDict.get(n, [])
    # nếu không còn màu trong danh sách (hiếm khi xảy ra với 4 màu ban đầu), thêm màu mới
    if len(available_colors) == 0:
        # thêm màu mới dạng Color_k để đảm bảo tồn tại màu
        new_color = f"Color{len(theSolution)+1}"
        available_colors = [new_color]
        colorDict[n] = available_colors

    # chọn màu đầu tiên có sẵn
    chosen_color = available_colors[0]
    theSolution[n] = chosen_color   # gán màu cho đỉnh n

    # Lấy danh sách kề của đỉnh n theo ma trận G (mảng 0/1)
    adjacentNode = G[t_[n]]

    # Loại bỏ màu vừa dùng khỏi danh sách màu khả dụng của các đỉnh kề
    for j in range(len(adjacentNode)):
        if adjacentNode[j] == 1:
            neigh = node[j]                   # tên đỉnh kề
            # nếu màu tồn tại trong danh sách màu khả dụng của đỉnh kề thì xóa
            if chosen_color in colorDict.get(neigh, []):
                colorDict[neigh].remove(chosen_color)

# In kết quả: theo thứ tự chữ cái (A..F) in ra màu đã tô
print("Coloring result (vertex : color):")
for v, c in sorted(theSolution.items()):
    print(f"{v} : {c}")
```

```python
# ------------------ BÀI 1 ---------------------

# Hàm đọc ma trận kề từ file txt
def read_adj_matrix(filename):
    G = []
    with open(filename, 'r') as f:
        for line in f:
            row = list(map(int, line.strip().split()))
            G.append(row)
    return G

# Thuật toán tô màu
def color_graph(G):
    n = len(G)
    node = [chr(ord('A') + i) for i in range(n)]  # Tạo tên A, B, C...

    # Bậc các đỉnh
    degree = [sum(G[i]) for i in range(n)]

    # Mỗi đỉnh có danh sách màu ban đầu
    colorDict = {node[i]: ["Blue", "Red", "Yellow", "Green"] for i in range(n)}

    # Sắp xếp theo bậc giảm dần (selection sort)
    sortedNode = []
    used = []
    for _ in range(n):
        max_deg = -1
        idx = -1
        for j in range(n):
            if j not in used and degree[j] > max_deg:
                max_deg = degree[j]
                idx = j
        used.append(idx)
        sortedNode.append(node[idx])

    # Ánh xạ tên → chỉ số
    index_map = {node[i]: i for i in range(n)}

    # Tô màu
    solution = {}
    for v in sortedNode:
        available_colors = colorDict[v]
        chosen_color = available_colors[0]  # màu đầu tiên
        solution[v] = chosen_color

        # Loại màu khỏi láng giềng
        for j in range(n):
            if G[index_map[v]][j] == 1:
                if chosen_color in colorDict[node[j]]:
                    colorDict[node[j]].remove(chosen_color)

    return solution

# ------------------ CHẠY CHƯƠNG TRÌNH ---------------------
file = "graph.txt"
G = read_adj_matrix(file)
result = color_graph(G)

print("Kết quả tô màu:")
for v, c in sorted(result.items()):
    print(v, "→", c)
```

```python
# ------------------ BÀI 2 ---------------------

def read_adj_matrix(filename):
    G = []
    with open(filename, 'r') as f:
        for line in f:
            row = list(map(int, line.split()))
            G.append(row)
    return G

def calculate_degrees(G):
    return [sum(row) for row in G]

def sort_nodes_by_degree(node, degree):
    sorted_nodes = []
    used = []
    n = len(node)

    for _ in range(n):
        max_deg = -1
        idx = -1
        for i in range(n):
            if i not in used and degree[i] > max_deg:
                max_deg = degree[i]
                idx = i
        used.append(idx)
        sorted_nodes.append(node[idx])

    return sorted_nodes

def color_graph(G):
    n = len(G)
    node = [chr(ord('A') + i) for i in range(n)]
    degree = calculate_degrees(G)
    sorted_nodes = sort_nodes_by_degree(node, degree)
    index_map = {node[i]: i for i in range(n)}

    colorDict = {node[i]: ["Blue", "Red", "Yellow", "Green"] for i in range(n)}
    solution = {}

    for v in sorted_nodes:
        chosen = colorDict[v][0]
        solution[v] = chosen

        # Loại màu khỏi đỉnh kề
        for j in range(n):
            if G[index_map[v]][j] == 1:
                if chosen in colorDict[node[j]]:
                    colorDict[node[j]].remove(chosen)

    return solution


def print_coloring(solution):
    print("\n--- KẾT QUẢ TÔ MÀU ---")
    for v, c in sorted(solution.items()):
        print(f"{v} → {c}")


# ------------------ MAIN ---------------------

file = "graph.txt"
G = read_adj_matrix(file)
solution = color_graph(G)
print_coloring(solution)
```
